#1. Forward Propagation
Definition

Forward Propagation is the process of passing the input data through the neural network to generate the final prediction (output).

It is called forward because the information moves only in one direction:

Input Layer → Hidden Layer(s) → Output Layer

No learning happens here.

The model simply makes a prediction using the current weights and biases.

Simple Real-Life Example

Imagine a student writing an exam.

Input:

Study Hours

Sleep Hours

Attendance

↓

↓

Prediction:

Pass /
Fail

The student's brain processes the information before giving the final decision.

Similarly,

Neural Network

Input → Hidden Layers → Prediction


##Flow of Forward Propagation
Input->
Weighted Sum (Z)->
Activation Function->
Output Prediction

#2. Loss Functions

#3. Backward Propagation
Definition

Backward Propagation is the learning process of a neural network. It calculates how much each weight and bias contributed to the error, then computes gradients so the model knows how to adjust them.

The information flows from the output layer back toward the input layer, which is why it is called backward propagation.

The objective is to reduce the loss by updating the weights in the correct direction.

Real-Life Example

Imagine a student writes an exam and scores 40 out of 100.

The teacher reviews the paper and identifies the mistakes:

Chapter 1 mistakes: 5 marks lost
Chapter 2 mistakes: 20 marks lost
Chapter 3 mistakes: 35 marks lost

The student now knows which topics caused the most errors and focuses on improving those topics.

Similarly, a neural network identifies which weights contributed most to the prediction error and adjusts those weights.

##Flow of Backward Propagation
Prediction->
Calculate Loss->
Compute Gradients->
Update Weights & Biases->
Lower Loss

###Why is Backward Propagation Needed?

Forward propagation only generates predictions.

Backward propagation answers these questions:

- How wrong is the prediction?
- Which weights caused the error?
- How should each weight change to reduce the error?

Without backward propagation, the network would never learn.

###The Role of Gradients

A gradient is the derivative of the loss with respect to a parameter (weight or bias). It tells us:

Direction: Should the weight increase or decrease?
Magnitude: By how much should it change?

Mathematically:
∂L/
∂W

Where:

L = Loss,
W = Weight

Weight Update Formula
Wnew=Wold−η(∂L/∂W)

Where:

η = Learning Rate
∂L/∂W= Gradient

If the gradient is positive, the weight decreases.

If the gradient is negative, the weight increases.

Example

Suppose:

Current Weight = 2.5

Gradient = 0.8

Learning Rate = 0.1

Updated Weight:2.5−(0.1×0.8)=2.42

The weight is adjusted slightly to reduce future errors.


```python
# Initial values
weight = 2.5
gradient = 0.8
learning_rate = 0.1

# Weight update
new_weight = weight - learning_rate * gradient

print("Old Weight:", weight)
print("Gradient:", gradient)
print("New Weight:", new_weight)
```

#4. Gradient Descent
Definition

Gradient Descent is an optimization algorithm used to minimize the loss function by updating the weights and biases in the direction that reduces the error.

Backward propagation computes the gradients, and Gradient Descent uses those gradients to update the parameters.

Real-Life Example

Imagine you are standing on top of a hill and want to reach the lowest point in the valley.

- The slope tells you which direction goes downhill.
- You take a small step.
- Check the slope again.
- Repeat until you reach the bottom.

Similarly:

- The loss function is the height of the hill.
- The gradient is the slope.
- The learning rate is the size of each step.
- Gradient Descent keeps taking steps until the loss is minimized.

Initialize Weights->
Forward Propagation->
Calculate Loss->
Backward Propagation->
Compute Gradients->
Gradient Descent Updates Weights->
Repeat Until Loss is Minimum

Effect of Learning Rate

Too Small: Training is very slow.

Too Large: The model may overshoot the minimum loss and fail to converge.

Optimal: Fast and stable convergence.


```python
# Initial values
weight = 5
gradient = 3
learning_rate = 0.1

for epoch in range(5):
    weight = weight - learning_rate * gradient
    print(f"Epoch {epoch+1}: Weight = {weight:.2f}")
```

#Building Neural Networks with TensorFlow/Keras

###What is TensorFlow?

TensorFlow is a tool (framework) that helps computers learn from data and make predictions.

It is developed by Google and is mainly used to build Machine Learning and Deep Learning models.

- Simple Definition

TensorFlow is a framework that helps us create, train, and use AI models.

- Real-Life Example

Imagine you want to teach a child to recognize cats and dogs.

You show many pictures of cats and dogs.
The child learns the differences.
After learning, the child can identify a new picture.

Similarly, TensorFlow helps a computer learn from examples and make predictions.

###What is Keras?

Keras is a library (API) that makes TensorFlow easy to use.

Instead of writing long and complicated code, Keras lets you build a neural network in just a few lines.

- Simple Definition

Keras is a simple interface that makes it easy to build deep learning models using TensorFlow.

- Real-Life Example

Think about driving a car.

 Car Engine → TensorFlow
 Steering Wheel → Keras

The engine (TensorFlow) does all the hard work.
The steering wheel (Keras) helps you control the car easily.

You don't directly control the engine—you use the steering wheel.

Similarly, programmers don't usually write low-level TensorFlow code. They use Keras, and Keras communicates with TensorFlow behind the scenes.

```python
# ==========================================
# Building a Neural Network using TensorFlow/Keras
# Example: Predict Pass/Fail based on Study Hours
# ==========================================

# Step 1: Import Libraries
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# ------------------------------------------
# Step 2: Create Dataset
# ------------------------------------------

# Input (Study Hours)
X = np.array([
    [2],
    [4],
    [6],
    [8],
    [10]
])

# Output (0 = Fail, 1 = Pass)
y = np.array([
    0,
    0,
    1,
    1,
    1
])

# ------------------------------------------
# Step 3: Build Neural Network
# ------------------------------------------

model = Sequential([
    Dense(8, activation='relu', input_shape=(1,)),   # Hidden Layer 1
    Dense(4, activation='relu'),                     # Hidden Layer 2
    Dense(1, activation='sigmoid')                   # Output Layer
])

# ------------------------------------------
# Step 4: Compile the Model
# ------------------------------------------

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# ------------------------------------------
# Step 5: Train the Model
# ------------------------------------------

print("Training Started...\n")

history = model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)

print("\nTraining Completed!")

# ------------------------------------------
# Step 6: Evaluate the Model
# ------------------------------------------

loss, accuracy = model.evaluate(X, y, verbose=0)

print("\nModel Evaluation")
print("-----------------------")
print("Loss     :", round(loss, 4))
print("Accuracy :", round(accuracy * 100, 2), "%")

# ------------------------------------------
# Step 7: Make Predictions
# ------------------------------------------

print("\nPredictions")
print("-----------------------")

test_data = np.array([
    [3],
    [5],
    [7],
    [9]
])

predictions = model.predict(test_data)

for hours, pred in zip(test_data, predictions):
    probability = pred[0]

    if probability >= 0.5:
        result = "Pass"
    else:
        result = "Fail"

    print(f"Study Hours: {hours[0]} -> Probability: {probability:.4f} -> Prediction: {result}")

# ------------------------------------------
# Step 8: Model Summary
# ------------------------------------------

print("\nModel Summary")
print("-----------------------")
model.summary()

# ------------------------------------------
# Step 9: Display Weights and Biases
# ------------------------------------------

print("\nModel Weights and Biases")
print("-----------------------")

for i, layer in enumerate(model.layers):
    weights, biases = layer.get_weights()

    print(f"\nLayer {i+1}")
    print("Weights:")
    print(weights)

    print("Biases:")
    print(biases)

# ------------------------------------------
# Step 10: Predict for a New Student
# ------------------------------------------

study_hours = 3

prediction = model.predict(np.array([[study_hours]]))

print("\nNew Student Prediction")
print("-----------------------")
print("Study Hours :", study_hours)
print("Probability :", prediction[0][0])

if prediction[0][0] >= 0.5:
    print("Final Prediction : PASS")
else:
    print("Final Prediction : FAIL")
```
