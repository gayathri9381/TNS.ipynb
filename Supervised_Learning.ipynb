## Supervised Learning

Definition:

Supervised Learning is a type of Machine Learning in which the model learns from labeled data. Every training example contains:

- Input (Features/Independent Variables - X)
- Output (Target/Dependent Variable - Y)

The algorithm learns the relationship between the input and output so that it can predict the output for new, unseen data.

-----------------------------------------------


---------------------------------------------------------

Simple Example

Suppose we want to predict whether a student will pass or fail.

Study Hours	Result
2	Fail
3	Fail
5	Pass
7	Pass
9	Pass

Here,

Input (X) = Study Hours
Output (Y) = Pass/Fail

The model learns this relationship and predicts the result for a student who studies for 6 hours.

Real-Life Analogy

Imagine a teacher teaching children.

The teacher shows many examples.

Cat  → Cat,
Dog  → Dog,
Cow  → Cow

After seeing many labeled examples, the child can identify a new animal correctly.

Similarly,

Teacher = Training Data,
Student = Machine Learning Model,
Learning = Training Process.

-----------------------------------------------

###Workflow of Supervised Learning

Collect Data

      ↓

Label the Data

      ↓

Split into Training & Testing Data

      ↓
Train the Model

      ↓
Evaluate the Model

      ↓
Predict New Data

----------------------------------------------------


####Advantages
- High prediction accuracy
- Easy to evaluate
- Suitable when labeled data is available
- Widely used in industry
####Disadvantages
- Requires large labeled datasets
- Labeling data is expensive
- Cannot discover hidden patterns without labels

Types of Supervised Learning

Supervised Learning is divided into two major types:

  1. Regression
  2. Classification

##Simple Linear Regression (Regression)
What is it?

Simple Linear Regression is used to predict a continuous value using one independent variable.

Formula -->
y=mx+c

Where

x → Independent Variable,
y → Dependent Variable,
m → Slope,
c → Intercept


```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
df = pd.DataFrame({
    "Experience":[1,2,3,4,5],
    "Salary":[3,4,5,6,7]
})

# Independent Variable
X = df[["Experience"]]

# Dependent Variable
y = df["Salary"]

# Create Model
model = LinearRegression()

# Train Model
model.fit(X,y)

# Prediction
prediction = model.predict([[6]])

print("Predicted Salary:",prediction[0])

print("Slope:",model.coef_[0])
print("Intercept:",model.intercept_)
```

##Multiple Linear Regression

What is it?

Multiple Linear Regression predicts the output using more than one input feature.

Formula
Y=b0+b1X1+b2X2+...+bnXn
	​

Example

Predict House Price

Inputs

Area,
Bedrooms,
Age,

Output

House Price

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    "Area":[1000,1200,1500,1800,2000],
    "Bedrooms":[2,2,3,3,4],
    "Age":[10,8,5,2,1],
    "Price":[40,45,60,75,90]
})

X = df[["Area","Bedrooms","Age"]]
y = df["Price"]

model = LinearRegression()

model.fit(X,y)

prediction = model.predict([[1700,3,4]])

print("Predicted Price:",prediction[0])
```

##Decision Tree

A Decision Tree makes decisions using a sequence of if-else conditions, similar to a flowchart.

Example

Loan Approval

Income > 50,000?

        Yes
         |
 Credit Score >700?

     Yes      No
      |        |
 Approve    Reject

No
|
Reject

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

df = pd.DataFrame({
    "Income":[25,35,45,60,70,80],
    "Loan":[0,0,0,1,1,1]
})

X = df[["Income"]]
y = df["Loan"]

model = DecisionTreeClassifier()

model.fit(X,y)

prediction = model.predict([[65]])

print("Loan Approved:",prediction[0])
```

##Logistic Regression (Classification)
What is Logistic Regression?

Logistic Regression is a Supervised Machine Learning Classification algorithm used to predict categorical outcomes, especially Binary Classification.

Although its name contains "Regression", it is used for Classification, not Regression.


#### Why is it called Logistic Regression?

Unlike Linear Regression, which predicts continuous values, Logistic Regression predicts the probability of belonging to a class.

It uses the Sigmoid Function to convert any value into a probability between 0 and 1.

```python

# Step 1: Import Libraries
import pandas as pd
from sklearn.linear_model import LogisticRegression

# Step 2: Create Dataset
df = pd.DataFrame({
    "Hours":[1,2,3,4,5,6],
    "Pass":[0,0,0,1,1,1]
})

print("Dataset")
print(df)

# Step 3: Input and Output
X = df[["Hours"]]
y = df["Pass"]
# Step 4: Create Model
model = LogisticRegression()
# Step 5: Train Model
model.fit(X,y)
# Step 6: Prediction
prediction = model.predict([[3.5]])
# Step 7: Output
print("\nPrediction:",prediction[0])

if prediction[0]==1:
    print("Student Passed")
else:
    print("Student Failed")
```

##Naive Bayes Classification
What is Naive Bayes?

Naive Bayes is a Supervised Machine Learning Classification algorithm based on Bayes' Theorem.

It predicts the class having the highest probability.

It is widely used for

Spam Detection,
Sentiment Analysis,
Email Classification,
Medical Diagnosis.

Bayes' Theorem P(A∣B)=P(B)P(B∣A)/P(A)
- P(A|B) → Probability of A given B
- P(B|A) → Probability of B given A
- P(A) → Prior Probability
- P(B) → Evidence

The algorithm calculates the probability for each class and predicts the class with the highest probability.
	​


Real-Life Example

Suppose an email contains these words:

Congratulations!

You Won Free Prize

Naive Bayes calculates

Probability(Spam) = 0.95

Probability(Not Spam) = 0.05

```python

# Step 1: Import Libraries
import pandas as pd
from sklearn.naive_bayes import GaussianNB

# Step 2: Create Dataset
df = pd.DataFrame({
    "Marks":[35,40,45,60,70,80],
    "Pass":[0,0,0,1,1,1]
})

print("Dataset")
print(df)

# Step 3: Input and Output
X = df[["Marks"]]
y = df["Pass"]

# Step 4: Create Model
model = GaussianNB()

# Step 5: Train Model
model.fit(X,y)

# Step 6: Prediction
prediction = model.predict([[65]])

# Step 7: Output
print("\nPrediction:",prediction[0])

if prediction[0]==1:
    print("Student Passed")
else:
    print("Student Failed")
```

**Data Visualization for SVM**

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.svm import SVC

# Create dataset
data = {
    "Hours": [1, 2, 3, 4, 5, 6],
    "Result": [0, 0, 0, 1, 1, 1]
}

df = pd.DataFrame(data)

# Input and Output
X = df[["Hours"]]
y = df["Result"]

# Train SVM model
model = SVC(kernel='linear')
model.fit(X, y)

# Predict values for plotting
x_values = np.linspace(0, 7, 100).reshape(-1, 1)
y_pred = model.predict(x_values)

# Plot original data
plt.scatter(df["Hours"], df["Result"], color="blue", label="Data Points")

# Plot prediction line
plt.plot(x_values, y_pred, color="red", label="SVM Prediction")

# Labels
plt.xlabel("Hours Studied")
plt.ylabel("Result")
plt.title("SVM Data Visualization")
plt.legend()

plt.show()
```

**Polynomial Regression**

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

data = {
    "Hours": [1, 2, 3, 4, 5, 6],
    "Marks": [10, 20, 35, 55, 80, 110]
}

df = pd.DataFrame(data)

X = df[["Hours"]]
y = df["Marks"]

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

prediction = model.predict(poly.transform([[4.5]]))

print("Predicted Marks:", prediction)
```

**Data Visualization for Polynomial Regression**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

data = {
    "Hours": [1, 2, 3, 4, 5, 6],
    "Marks": [10, 20, 35, 55, 80, 110]
}

df = pd.DataFrame(data)

# Input and Output
X = df[["Hours"]]
y = df["Marks"]

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

predicted = model.predict(X_poly)

plt.bar(df["Hours"] - 0.2, y, width=0.4, label="Actual Marks")
plt.bar(df["Hours"] + 0.2, predicted, width=0.4, label="Predicted Marks")

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Actual vs Predicted Marks")
plt.legend()

plt.show()
```

**Write a program on detecting whether the mail is spam or not**

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

data = {
    "Mail": [
        "Win a free mobile",
        "Claim your prize now",
        "Meeting at 10 AM",
        "Project submission tomorrow"
    ],
    "Type": [
        "Spam",
        "Spam",
        "Ham",
        "Ham"
    ]
}

df = pd.DataFrame(data)

X = df["Mail"]
y = df["Type"]

# Convert text into numbers
cv = CountVectorizer()
X = cv.fit_transform(X)

# Train the model
model = MultinomialNB()
model.fit(X, y)

# Test mail
test_mail = ["Win a free mobile"]

# Convert test mail into numbers
test = cv.transform(test_mail)

# Predict
result = model.predict(test)

print("Prediction:", result[0])
```
